## 400-2 — 새 행 추가 후 loc·iloc 결과 비교

**문제**

`{"Name": "신규상장A", "Market": "KOSPI", "Sector": "IT서비스", "Marcap": 5_000_000_000_000}`을 인덱스 `"999999"`로 하는 한 행짜리 DataFrame을 만들어 `pd.concat`으로 `df_indexed` 맨 앞에 붙이고 `df_indexed2`에 저장하세요. `df_indexed2.loc['005930', 'Name']`와 `df_indexed2.iloc[0]['Name']`을 출력하세요.

**작성 위치**: 바로 아래 코드 셀에 위 문제의 코드를 입력하세요. instructor와 student는 같은 문제를 공유합니다.

[→ §1 강의노트](../notes/FD-02-400__indexing-filter.md)

```python
# 📌 §400 cell_id=400:2 실습
# 코드를 작성해주세요 👇
# 과제: 강의노트의 입력과 같은 결과가 나오도록 한 단계씩 작성한다.
new_df=pd.DataFrame(
    [{"Name": "신규상장A", "Market": "KOSPI", "Sector": "IT서비스", "Marcap": 5_000_000_000_000}]
    ,index=["999999"])
df_indexed2=pd.concat([new_df,df_indexed])

print(df_indexed2.loc['005930':'005935','Name'])
print(df_indexed2.iloc[0]['Name'])
# 예상 출력:
# 실행 후 강의노트의 결과와 비교하세요.
```
