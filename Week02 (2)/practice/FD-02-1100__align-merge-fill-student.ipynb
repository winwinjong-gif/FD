## 500-7 — inner join

**문제**

`samsung`과 `usdkrw`의 `Close` 컬럼을 각각 `Samsung`, `USDKRW`로 이름 바꿔 날짜 인덱스 기준 `how='inner'`로 합치고, `shape`와 `isnull().sum()`을 출력하세요.

**작성 위치**: 바로 아래 코드 셀에 위 문제의 코드를 입력하세요. instructor와 student는 같은 문제를 공유합니다.

[→ §5 강의노트](../notes/FD-02-500__align-merge-fill.md)

```python
# 📌 §500-7 실습
# 코드를 작성해주세요 👇
# 과제: 강의노트의 입력과 같은 결과가 나오도록 한 단계씩 작성한다.
# 힌트: 바로 위의 설명과 고정 CSV의 열 이름을 다시 확인한다.
s=samsung[['Close']].rename(columns={'Close':'Samsung'})
n=naver[['Close']].rename(columns={'Close':'Naver'})
inner=pd.merge(s,n,left_index=True,right_index=True,how="inner")
print(inner.shape)
print(inner.isnull().sum())
# 예상 출력:
# 실행 후 강의노트의 결과와 비교하세요.
```
