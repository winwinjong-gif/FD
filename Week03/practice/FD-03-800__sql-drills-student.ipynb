## ➕ 없는 열을 만들어 쓴다 — AS · 계산 컬럼

이 예시의 실행 흐름을 먼저 예측한 뒤, 같은 질문을 노트북에서 확인한다.

[→ §300 강의노트](../notes/FD-03-300__sql-remind.md)


```python
# 📌 §300 ➕ 없는 열을 만들어 쓴다 — AS · 계산 컬럼 cell_id=300:4
# 코드를 작성해주세요 👇
query="""
select code, net_income, equity, net_income/equity as roe_wrong,
round(net_income*1.0/equity,4) as roe_right 
from financials where year==2025
limit 5
"""
pd.read_sql(query,conn)
# ✍️ 힌트: 분자에 1.0을 곱해 소수 나눗셈을 만든다.
# 예상 출력:
# 강의노트의 실행 결과와 비교한다.
```

```python
# 📌 §300 🧭 쉬운 한 표로 먼저 묻기 — WHERE · ORDER BY · LIMIT cell_id=300:13
# 코드를 작성해주세요 👇
query =""" 
select code, revenue from financials where year==2025 and revenue>300000
order by revenue desc limit 5
"""

pd.read_sql(query,conn)
# ✍️ 힌트: financials 한 표에서 2025년 매출 300000 초과 행만 남기고, 매출 내림차순으로 5개를 고른다.
# 예상 출력:
# 강의노트의 실행 결과와 비교한다.
```

```python
query=""" 
select * from companies limit 5"""

pd.read_sql(query,conn)
```

```python
# 📌 §300 🔒 심화/선택 — 산업별로 묶는다 — GROUP BY · 집계 cell_id=300:5
# 코드를 작성해주세요 👇
query="""
select c.code, f.revenue, c.sector, round(avg(f.net_income*1.0/f.equity),4) as avg_roe,count(*) as cnt
from companies c join financials f on c.code=f.code
where f.year=2025
group by c.sector
order by avg_roe desc
limit 5
"""
pd.read_sql(query,conn)
# ✍️ 힌트: 앞의 집계 결과를 avg_roe 내림차순으로 정렬한다.
# 예상 출력:
# 강의노트의 실행 결과와 비교한다.
```

```python
# 📌 §300 🔒 심화/선택 — 미션 1 — 특정 업종에서 매출 상위 5개 회사 cell_id=300:10
# 코드를 작성해주세요 👇
query=""" 
select f.code, f.revenue from companies c
join financials f on c.code==f.code 
where c.sector="전기전자"
order by f.revenue desc limit 5
 """
pd.read_sql(query,conn)
# ✍️ 힌트: companies와 financials를 code로 잇고, WHERE로 업종을 고른 뒤 ORDER BY ... DESC LIMIT 5.
# 예상 출력:
# 강의노트의 실행 결과와 비교한다.
```

## ✍️ 미션 2 — 업종별 평균 수익성과 회사 수

이 예시의 실행 흐름을 먼저 예측한 뒤, 같은 질문을 노트북에서 확인한다.

[→ §300 강의노트](../notes/FD-03-300__sql-remind.md)


```python
# 📌 §300 🔒 심화/선택 — 미션 2 — 업종별 평균 수익성과 회사 수 cell_id=300:11
# 코드를 작성해주세요 👇
query=""" 
select c.sector, round(avg(f.net_income*1.0/f.equity),4) as avg_roe, count(*) as n
from companies c join financials f on c.code=f.code
where f.year=2025
group by c.sector
order by avg_roe desc
limit 5
"""
pd.read_sql(query,conn)
# ✍️ 힌트: 위 GROUP BY 절에서 쓴 쿼리와 같은 모양이다. 묶은 기준(sector)과 집계값(AVG, COUNT)만 고른다.
# 예상 출력:
# 강의노트의 실행 결과와 비교한다.
```
