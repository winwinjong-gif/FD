```python
# 📌 §400 🔍 받아서 이어 한다 cell_id=400:5
# 코드를 작성해주세요 👇
query=""" 
select round(net_income*1.0/equity,4) as roe from financials"""
df=pd.read_sql(query,conn)
mask=df["roe"]>0.08
print(df[mask].sort_values("roe",ascending=False))
# ✍️ 힌트: df에서 roe가 0.08보다 큰 행만 남기고 roe 내림차순 정렬한다.
# 예상 출력:
# 강의노트의 실행 결과와 비교한다.
```
